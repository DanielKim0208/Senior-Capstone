### Mobile TX effectiveness calculation via geometric analysis

Prior work on wireless power transfer has focused on fixed ground stations, rotating ground stations, or beamforming. The calculation below shows how lateral movement of the transmitter increases effective coverage.
 

```python
import numpy as np
import ipywidgets as widgets
from IPython.display import display

def make_slider(description, value=0.5, vmin=0, vmax=1, step=0.01):
    return widgets.FloatSlider(
        value=value, min=vmin, max=vmax, step=step,
        description=description,
        continuous_update=False,
        readout_format='.3f',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='500px'),
    )

def effectivenessCalculations(tx, actuator):
    rotational = tx ** 2 * np.pi
    mobile = tx ** 2 * np.pi + 2 * tx * actuator
    pct_increase = (mobile - rotational) / rotational * 100 if rotational > 0 else float('nan')
    return rotational, mobile, pct_increase

tx_slider = make_slider('Effective TX radius:', vmin=0.05)   # vmin > 0 avoids the divide-by-zero region
actuator_slider = make_slider('Actuator length:')            # adjust range/label as needed

out = widgets.Output()

def on_change(change=None):
    rotational, mobile, pct = effectivenessCalculations(tx_slider.value, actuator_slider.value)
    with out:
        out.clear_output(wait=True)
        print(f"Rotational: {rotational:.3f} m^2 ")
        print(f"Mobile:     {mobile:.3f} m^2")
        print(f"Increase:   {pct:.1f}%")

tx_slider.observe(on_change, names='value')
actuator_slider.observe(on_change, names='value')

on_change()  # print initial values
display(tx_slider, actuator_slider, out)
```
